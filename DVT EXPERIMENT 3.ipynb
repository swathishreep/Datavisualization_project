```python
# ============================================================
# GOOGLE PLAY STORE ANALYSIS
# DATA PREPROCESSING / DATA CLEANING
# SINGLE-RUN GOOGLE COLAB VERSION
# ============================================================

# -----------------------------
# 1. Import Libraries
# -----------------------------

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, LabelEncoder

from google.colab import files


# -----------------------------
# 2. Upload Dataset
# -----------------------------

print("Please upload your Google Play Store CSV file.")
print("Upload: googleplaystore.csv")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("\nUploaded:", filename)


# -----------------------------
# 3. Read Dataset
# -----------------------------

df = pd.read_csv(filename)

print("\nDataset loaded successfully.")

print("\nFirst 5 Rows:")
display(df.head())


# -----------------------------
# 4. Dataset Information
# -----------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# -----------------------------
# 5. Missing Values
# -----------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# -----------------------------
# 6. Convert Numerical Columns
# -----------------------------

# Reviews
df["Reviews"] = pd.to_numeric(
    df["Reviews"],
    errors="coerce"
)

# Installs
df["Installs"] = pd.to_numeric(
    df["Installs"]
    .astype(str)
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

# Price
df["Price"] = pd.to_numeric(
    df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False),
    errors="coerce"
)

# Rating
df["Rating"] = pd.to_numeric(
    df["Rating"],
    errors="coerce"
)

print("\nNumerical Columns Converted Successfully.")

print(
    df[
        ["Rating", "Reviews", "Installs", "Price"]
    ].dtypes
)


# -----------------------------
# 7. Handle Missing Values
# -----------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

print("\nMissing Values After Imputation:")
print(df.isnull().sum())


# -----------------------------
# 8. Remove Duplicate Rows
# -----------------------------

print(
    "\nDuplicate Rows:",
    df.duplicated().sum()
)

df.drop_duplicates(
    inplace=True
)

print(
    "After Duplicate Removal:",
    df.shape
)


# -----------------------------
# 9. Check Rating Outliers
# -----------------------------

Q1 = df["Rating"].quantile(0.25)

Q3 = df["Rating"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Rating"] < lower) |
    (df["Rating"] > upper)
]

print("\nRating Outliers:")
display(outliers.head())

print(
    "Number of Rating Outliers:",
    len(outliers)
)


# -----------------------------
# 10. Rating Boxplot
# -----------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["Rating"]
)

plt.title(
    "Google Play Store Rating Outliers"
)

plt.xlabel("Rating")

plt.show()


# -----------------------------
# 11. Remove Invalid Ratings
# -----------------------------

df = df[
    (df["Rating"] >= 1) &
    (df["Rating"] <= 5)
]

print(
    "\nAfter Removing Invalid Ratings:",
    df.shape
)


# -----------------------------
# 12. Min-Max Scaling
# -----------------------------

scaler = MinMaxScaler()

scale_cols = [
    "Reviews",
    "Installs",
    "Price"
]

df[scale_cols] = scaler.fit_transform(
    df[scale_cols]
)

print("\nScaled Values:")
display(
    df[
        ["Reviews", "Installs", "Price"]
    ].head()
)


# -----------------------------
# 13. Label Encoding - Type
# -----------------------------

type_encoder = LabelEncoder()

df["Type"] = type_encoder.fit_transform(
    df["Type"].astype(str)
)

print("\nEncoded Type:")
display(
    df[["Type"]].head()
)


# -----------------------------
# 14. Label Encoding - Content Rating
# -----------------------------

content_encoder = LabelEncoder()

df["Content Rating"] = content_encoder.fit_transform(
    df["Content Rating"].astype(str)
)

print("\nEncoded Content Rating:")
display(
    df[["Content Rating"]].head()
)


# -----------------------------
# 15. One-Hot Encoding - Category
# -----------------------------

category = pd.get_dummies(
    df["Category"],
    prefix="Category"
)

print("\nEncoded Category:")
display(
    category.head()
)


# -----------------------------
# 16. One-Hot Encoding - Genres
# -----------------------------

genres = pd.get_dummies(
    df["Genres"],
    prefix="Genre"
)

print("\nEncoded Genres:")
display(
    genres.head()
)


# -----------------------------
# 17. Combine Encoded Data
# -----------------------------

df = pd.concat(
    [
        df.drop(
            ["Category", "Genres"],
            axis=1
        ),
        category,
        genres
    ],
    axis=1
)

print(
    "\nFinal Dataset Shape:",
    df.shape
)

print("\nFinal Dataset:")
display(df.head())


# -----------------------------
# 18. Final Missing Value Check
# -----------------------------

print("\nTotal Remaining Missing Values:")

print(
    df.isnull().sum().sum()
)


# -----------------------------
# 19. Save Cleaned Dataset
# -----------------------------

output_file = "cleaned_googleplaystore.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    "\nCleaned dataset saved successfully as:",
    output_file
)


# -----------------------------
# 20. Download Cleaned Dataset
# -----------------------------

files.download(
    output_file
)

print("\n============================================")
print("DATA PREPROCESSING COMPLETED SUCCESSFULLY")
print("============================================")
```
